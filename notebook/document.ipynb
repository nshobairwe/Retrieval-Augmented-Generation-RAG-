```python
sample_texts = {
    "../data/text_files/python_intro.txt":"""PYTHON AND MACHINE LEARNING
============================

1. What is Python?
------------------
Python is a high-level, general-purpose programming language known for its
simple syntax and readability. It is widely used in web development, data
analysis, automation, artificial intelligence, and machine learning.

Example:

    name = "Witness"
    print("Hello", name)

Why Python is useful for Machine Learning:
- It is easy to learn and write.
- It has many powerful libraries.
- It has a large developer community.
- It supports data processing and visualization.
- It works well with popular machine learning frameworks.

Important Python libraries:
- NumPy: numerical and mathematical operations.
- Pandas: data manipulation and analysis.
- Matplotlib: data visualization.
- Scikit-learn: traditional machine learning.
- TensorFlow: deep learning.
- PyTorch: deep learning and AI research.
    "../data/text_files/machine_learning_intro.txt":"

2. What is Machine Learning?
----------------------------
Machine Learning (ML) is a branch of Artificial Intelligence (AI) where
computers learn patterns from data and use those patterns to make
predictions or decisions without being explicitly programmed for every
possible situation.

Simple example:

Suppose we have information about houses:

    Size (m²)    Price
    50           50,000
    80           80,000
    100          100,000
    150          150,000

A machine learning model can learn the relationship between house size
and price. After training, it can estimate the price of a new house.


3. Main Types of Machine Learning
---------------------------------

A. Supervised Learning
The model learns from data that already has known answers (labels).

Examples:
- Predicting house prices.
- Detecting fraudulent transactions.
- Predicting whether an email is spam.

Common algorithms:
- Linear Regression
- Logistic Regression
- Decision Tree
- Random Forest
- Support Vector Machine (SVM)
- K-Nearest Neighbors (KNN)


B. Unsupervised Learning
The data does not contain predefined labels. The model tries to discover
patterns or groups in the data.

Examples:
- Grouping customers by behavior.
- Finding similar products.
- Customer segmentation.

Common algorithms:
- K-Means Clustering
- Hierarchical Clustering
- PCA


C. Reinforcement Learning
An agent learns by interacting with an environment. It receives rewards
for good actions and penalties for bad actions.

Examples:
- Game-playing agents.
- Robot control.
- Autonomous decision-making systems.


4. Typical Machine Learning Workflow
------------------------------------
A basic machine learning project usually follows these steps:

1. Collect data.
2. Understand the data.
3. Clean the data.
4. Prepare and transform the data.
5. Split the data into training and testing sets.
6. Select a machine learning algorithm.
7. Train the model.
8. Evaluate the model.
9. Improve or tune the model.
10. Deploy the model for real-world use.


5. Example Using Python
-----------------------
A simple supervised learning example using Scikit-learn:

    from sklearn.linear_model import LinearRegression

    X = [[50], [80], [100], [150]]
    y = [50000, 80000, 100000, 150000]

    model = LinearRegression()
    model.fit(X, y)

    prediction = model.predict([[120]])
    print(prediction)

Here:
- X contains the input features (house size).
- y contains the target values (house prices).
- fit() trains the model.
- predict() uses the trained model to make a prediction.


6. Important Machine Learning Concepts
--------------------------------------
Feature:
An input variable used by a model.

Example:
For house-price prediction, size, location, and number of bedrooms can
be features.

Label/Target:
The value that the model is trying to predict.

Training Data:
Data used to teach the model.

Testing Data:
Data used to evaluate how well the trained model performs on unseen data.

Model:
A mathematical representation learned from data.

Overfitting:
When a model learns the training data too closely and performs poorly
on new data.

Underfitting:
When a model is too simple to learn the important patterns in the data.

Accuracy:
The proportion of correct predictions, commonly used for classification.

Precision and Recall:
Metrics commonly used to evaluate classification models, especially when
the classes are imbalanced.

Mean Squared Error (MSE):
A common metric for measuring prediction errors in regression.

R²:
A regression metric that indicates how much of the variation in the target
is explained by the model.


7. Python + Machine Learning
----------------------------
Python provides the programming environment, while machine learning
libraries provide tools for building models.

A common learning path is:

    Python
       |
       v
    NumPy + Pandas
       |
       v
    Data Cleaning & Visualization
       |
       v
    Statistics
       |
       v
    Scikit-learn
       |
       v
    Machine Learning
       |
       v
    Deep Learning / AI


8. Key Idea to Remember
-----------------------
Python is a programming language.

Machine Learning is a method of teaching computers to learn patterns
from data.

In practice:

    Data + Algorithm + Training = Machine Learning Model

The model can then be used to make predictions or decisions on new data."""
}

for filepath,content in sample_texts.items():
    with open(filepath, 'w', encoding = 'utf-8') as f:
        f.write(content)

print("Sample text file created!")
```

```python
### Directory Loader
from langchain_community.document_loaders import DirectoryLoader

## Load all the text files in the directory
dir_loader = DirectoryLoader(
    "../data/text_files",
    glob = "**/*.txt", ## Pattern to match files
    loader_cls = TextLoader, ## Loader class to use base on the type of file you load
    loader_kwargs = {"encoding": "utf-8"},
    show_progress = False
)

documents =dir_loader.load()
documents
```
